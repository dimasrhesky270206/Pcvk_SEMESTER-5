TUGAS PRAKTIKUM - Aplikasi

Pada bagian ini, Anda hanya diperbolehkan memakai fungsi yang telah dipelajari pada modul ini: pembacaan berkas, pengirisan, cv2.resize, cv2.flip, cv2.cvtColor, fungsi penggambaran OpenCV, serta pembuatan kanvas dengan NumPy. Selain itu, program harus bersifat adaptif, dimana setiap ukuran, posisi, atau ketebalan harus dihitung dari ukuran citra masukan, bukan ditulis sebagai angka tetap pada kode program.

Studi Kasus 1: Panitia sebuah kegiatan perlu membubuhkan label identitas pada ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel, sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama.

jawab:




```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tambah_label_identitas(img):
    # 1. Ambil dimensi citra (tinggi, lebar, kanal)
    h, w = img.shape[:2]

    # Buat salinan agar citra asli tidak tertimpa
    img_out = img.copy()

    # 2. Hitung dimensi bar secara adaptif (misal: tinggi bar 10% dari tinggi citra)
    bar_height = int(h * 0.10)

    # 3. Buat bar transparan/solid di bagian bawah citra
    # Koordinat pt1 (kiri-atas bar) dan pt2 (kanan-bawah bar)
    pt1 = (0, h - bar_height)
    pt2 = (w, h)
    cv2.rectangle(img_out, pt1, pt2, color=(0, 0, 0), thickness=-1)

    # 4. Hitung skala font, ketebalan, dan posisi teks secara adaptif
    # Skala font disesuaikan dengan lebar citra
    font_scale = w / 800.0  # Basis proporsi untuk skala 1.0 pada lebar 800px
    font_thickness = max(1, int(w / 400))

    teks = "DOKUMENTASI KEGIATAN - 2026"
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Hitung ukuran teks dalam piksel untuk pemosisian presisi (centered)
    (text_w, text_h), baseline = cv2.getTextSize(teks, font, font_scale, font_thickness)

    # Posisi teks di tengah bar bawah
    text_x = int((w - text_w) / 2)
    text_y = int(h - (bar_height - text_h) / 2)

    # 5. Tuliskan teks label pada bar
    cv2.putText(img_out, teks, (text_x, text_y), font, font_scale, (255, 255, 255), font_thickness)

    return img_out

# === PENGUJIAN PADA DUA CITRA BERBEDA UKURAN ===

# 1. Baca citra asli (misal: Citra A)
img_a = cv2.imread('//content/drive/MyDrive/pcvk/ktm.png')
img_a = cv2.cvtColor(img_a, cv2.COLOR_BGR2RGB)

# 2. Buat Citra B dengan ukuran berbeda (misal: 50% dari ukuran Citra A menggunakan cv2.resize)
h_orig, w_orig = img_a.shape[:2]
img_b = cv2.resize(img_a, (int(w_orig * 0.5), int(h_orig * 0.5)))

# 3. Jalankan fungsi adaptif pada kedua citra
hasil_a = tambah_label_identitas(img_a)
hasil_b = tambah_label_identitas(img_b)

# 4. Tampilkan kedua hasil secara berdampingan
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(hasil_a)
axes[0].set_title(f'Citra A (Ukuran Besar: {img_a.shape[1]}x{img_a.shape[0]})')
axes[0].axis('off')

axes[1].imshow(hasil_b)
axes[1].set_title(f'Citra B (Ukuran Kecil: {img_b.shape[1]}x{img_b.shape[0]})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
