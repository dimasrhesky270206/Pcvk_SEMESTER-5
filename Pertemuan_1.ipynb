```python
import numpy as np
import cv2 as cv
from skimage import io
from google.colab.patches import cv2_imshow

urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placecats.com/800/571"]

```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placecats.com/800/571"]

for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(lebar):
  image_3[int((tinggi)/2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)


# Membuat garis vertikal dengan loop (menggunakan pembagian bulat //)
for x in range(tinggi):
    image_3[x, lebar // 2] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Garis Diagonal (Menyilang)
y_coords = np.linspace(0, tinggi - 1, num=lebar).astype(int)
x_coords = np.arange(lebar)

# Diagonal dari kiri-atas ke kanan-bawah
image_3[y_coords, x_coords] = [255, 255, 255]

# Diagonal dari kiri-bawah ke kanan-atas
image_3[tinggi - 1 - y_coords, x_coords] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Ambil tinggi dan lebar gambar dari shape
tinggi, lebar = image.shape[:2]

# Hitung titik tengah vertikal (Y)
y_tengah = tinggi // 2

# Menentukan panjang garis yang diinginkan (100 piksel)
panjang_garis = 100

# Hitung koordinat awal (X1) dan akhir (X2) agar posisi garis presisi di tengah
start_x = max(0, (lebar - panjang_garis) // 2)
end_x = min(lebar, start_x + panjang_garis)

# Gambar garis putih pada rentang koordinat X tersebut
image_3[y_tengah, start_x:end_x] = [255, 255, 255]

# Gabungkan dan tampilkan gambar
final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_4 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_5 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

panjang = min(tinggi, lebar)
for i in range(panjang):
  image_5[i, i] = [255, 255, 255]
  image_5[i, lebar - 1 - i] = [255, 255, 255]

final_frame = cv.hconcat((image_4, image_5))
cv2_imshow(final_frame)

image_6 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_7 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

panjang_garis = lebar // 2
mulai = (lebar - panjang_garis) // 2
baris_tengah = tinggi // 2

for y in range(mulai, mulai + panjang_garis):
  image_7[baris_tengah, y] = [255, 255, 255]

final_frame = cv.hconcat((image_6, image_7))
cv2_imshow(final_frame)

image_8 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_9 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

ukuran_kotak = 60
pusat_baris = int(tinggi * 0.6)
pusat_kolom = int(lebar * 0.6)

for i in range(pusat_baris, pusat_baris + ukuran_kotak):
  for j in range(pusat_kolom, pusat_kolom + ukuran_kotak):
    image_9[i, j] = [255, 255, 255]

final_frame = cv.hconcat((image_8, image_9))
cv2_imshow(final_frame)
```
